```python
#bài w2A4
# Đọc 6 điểm số từ một dòng, cách nhau bởi dấu cách
a1, b1, c1, a2, b2, A3 = map(float, input().split())

# Tính điểm trung bình (TB) theo công thức
tb = (a1 + b1 + c1 + (a2 + b2) * 2 + A3 * 3) / 10

# In kết quả đã được làm tròn đến 1 chữ số thập phân
print(f"{tb:.2f}")
```
