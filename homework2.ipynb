```python
#bai5
import math
a=int(input('nhap a='))
b=int(input('nhap b='))
c=int(input('nhap c='))
print(a,'x^2 +',b,'x',c,'= 0')
denta=b**2-4*a*c
x1=0
x2=0
if denta>0:
    x1=(-b+ math.sqrt(denta))/(2*a)
    x2=(-b- math.sqrt(denta))/(2*a)
    print('pt co 2 nghiem')
    print('x1=',x1,'x2=',x2)
elif denta==0:
    x1=x2=-b/(2*a)
    print('pt co nghiem kep=',x1)
else:
    print('pt vo nghiem')
```
